```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_chroma import Chroma
from langchain_core.documents import Document
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
from langchain_core.documents import Document

# Create LangChain documents for football players

doc1 = Document(
    page_content="Lionel Messi is widely regarded as one of the greatest footballers of all time. Known for his incredible dribbling, vision, passing, and goal-scoring ability, he has achieved numerous trophies throughout his career.",
    metadata={"club": "Inter Miami"}
)

doc2 = Document(
    page_content="Cristiano Ronaldo is one of the most prolific goal scorers in football history. Known for his athleticism, powerful shooting, heading ability, and incredible work ethic, he has won numerous domestic and international trophies.",
    metadata={"club": "Al-Nassr"}
)

doc3 = Document(
    page_content="Kylian Mbappe is one of the most exciting young footballers in the world. Known for his explosive pace, dribbling, finishing, and ability to score in important matches, he has established himself as a world-class forward.",
    metadata={"club": "Real Madrid"}
)

doc4 = Document(
    page_content="Kevin De Bruyne is one of the best midfielders of his generation. He is known for his exceptional passing, vision, crossing, long-range shooting, and ability to create goal-scoring opportunities.",
    metadata={"club": "Manchester City"}
)

doc5 = Document(
    page_content="Virgil van Dijk is regarded as one of the best defenders in modern football. Known for his strength, positioning, aerial ability, leadership, and composure, he has been an important player for both club and country.",
    metadata={"club": "Liverpool"}
)
```

```python
docs = [doc1, doc2, doc3, doc4, doc5]
```

```python
embedding_model = GoogleGenerativeAIEmbeddings(
    model='models/gemini-embedding-001'
)

vector_store = Chroma(
    embedding_function=embedding_model,
    persist_directory="chroma_db",
    collection_name="simple"
)
```

```python
# Add documents
vector_store.add_documents(docs)
```

```python
# View documents
vector_store.get(include=['embeddings', 'documents', 'metadatas'])
```

```python
# Search documents
vector_store.similarity_search(
    query='Who among these are a shooter',
    k=2
)
```

```python
vector_store.similarity_search_with_score(
    query='Who among these are a shooter',
    k=2
)
```

```python
# Metadata filtering
vector_store.similarity_search_with_score(
    query="Which player plays for this club?",
    filter={"club": "Manchester City"},
)
```

```python
# Update existing documents
updated_doc2 = Document(
    page_content="Cristiano Ronaldo is renowned for his incredible speed, powerful finishing, aerial ability, and goal-scoring consistency. His dedication to training and competitive mentality have made him one of the most successful footballers of his generation.",
    metadata={"club": "Al-Nassr"}
)

vector_store.update_document(document_id='22f05d40-7b0d-410e-bf6b-436bf1d47b48', document=updated_doc2)
```

```python
vector_store.get(include=['embeddings', 'documents', 'metadatas'])
```

```python
# Delete document
vector_store.delete(ids=['22f05d40-7b0d-410e-bf6b-436bf1d47b48'])
```
