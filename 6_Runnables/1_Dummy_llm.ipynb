```python
import random
```

```python
class DummyLLM:
    def __init__(self):
        print("LLM has created")

    def predict(self, prompt):

        response_list = [
            "Old Dhaka biriyani is the best",
            "Tony Stark made jarvis",
            "I can be something else"
        ]

        return {'response': random.choice(response_list)}
```

```python
llm = DummyLLM()
```

```python
llm.predict('Say something')
```

```python
class DummyPromptTemplate:

    def __init__(self, template, input_variables):
        self.template = template
        self.input_variables = input_variables

    def format(self, input_dict):
        return self.template.format(**input_dict)
```

```python
template = DummyPromptTemplate(
    template='Write a {length} peom about {topic}',
    input_variables=['topic']
)
```

```python
prompt = template.format({'length':'short', 'topic':'Bangladesh'})
```

```python
llm.predict(prompt)
```

```python
class DummyLLMChain:
    def __init__(self, llm, prompt):
        self.llm = llm
        self.prompt = prompt

    def run(self, input_dict):
        final_prompt = self.prompt.format(input_dict)
        result = self.llm.predict(final_prompt)

        return result['response']
```

```python
template = DummyPromptTemplate(
    template='Write a {length} peom about {topic}',
    input_variables=['topic']
)
```

```python
llm = DummyLLM()
```

```python
chain = DummyLLMChain(llm, template)
```

```python
chain.run({'length':'short', 'topic':'Bangladesh'})
```
